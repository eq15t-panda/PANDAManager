```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
from utils.plot_style import set_plot_style
set_plot_style()
```

```python
# Import pyrpl data files (.npz)
pyrpl_folder = "data/pyrpl/"
pyrpl_files = [f for f in os.listdir(pyrpl_folder) if f.endswith('.npz')]
print("Available pyrpl data files:" +str(pyrpl_files))
```

```python
# Select a file to load
file_to_load = pyrpl_files[2]  # Change this index to load a different file
data = np.load(os.path.join(pyrpl_folder, file_to_load))

print(data.files)  # List all arrays in the .npz file
```

```python
# Extract ch1, ch2 and time data from oscilloscope
ch1 = data["ch1"]
ch2 = data["ch2"]
time = data["time"]

# Plot the data
plt.figure(figsize=(10, 6))
plt.plot(time, ch1, label='Channel 1')
plt.plot(time, ch2, label='Channel 2')
plt.xlabel('Time (s)')
plt.ylabel('Voltage (V)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Extract frequency and transfer function data
freq = data["freq"]
tf = data["tf"]

magnitude = 20*np.log10(np.abs(tf))
phase = np.angle(tf, deg=True)

# Magnitude
plt.figure(figsize=(7,4))
plt.semilogx(freq, magnitude)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude (dB)")
plt.grid(True)
plt.show()

# Phase
plt.figure(figsize=(7,4))
plt.semilogx(freq, phase)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Phase (deg)")
plt.grid(True)
plt.show()
```
