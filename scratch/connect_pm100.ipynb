```python
import pyvisa as pv 
rm = pv.ResourceManager()
print(rm.list_resources())
```

```python
pm_100 = rm.open_resource('USB0::0x1313::0x8079::P1007388::INSTR')
```

```python
pm_100.query('*IDN?')
```

```python
import numpy as np
a = pm_100.query( "READ?")
t = 2
import time 
tt = 0 
list_P = []
t0 = time.time()
while tt < t:
    a = pm_100.query( "READ?")
    list_P.append(float(a[:-1]))
    tt += time.time() - t0

print(np.mean(list_P))
```
